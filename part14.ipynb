# merge function

```python
import numpy as np
import pandas as pd
```

```python
match = pd.read_csv('matches.csv')
match.head()
```

```python
delivery = pd.read_csv('deliveries.csv')
delivery.head()
```

## Year by highest run or orange cap winne


```python
new = delivery.merge(match, left_on='match_id', right_on='id')
new.head()
```

### multi dimension series

```python
new.groupby(['season', 'batsman'])['batsman_runs'].sum()
```

```python
new.groupby(['season', 'batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index()
```

```python
new.groupby(['season', 'batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().drop_duplicates(subset='season', keep='first')
```

```python
cap = new.groupby(['season', 'batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().drop_duplicates(subset='season', keep='first')
cap
```
