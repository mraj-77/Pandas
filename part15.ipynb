# Pivot table

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('indian_food.csv')
df.head()
```

```python
# 1. Dataset mein '-1' (string) aur -1 (number) ko NaN se replace karein
df.replace(['-1', -1], np.nan, inplace=True)

# 2. Ab clean data par pivot table banayein (NaN automatically ignore ho jayenge)
df.pivot_table(index=['state', 'region'], columns='diet', values='cook_time', aggfunc='mean')
```
