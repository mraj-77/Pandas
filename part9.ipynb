# sort value method

# sort_values()

- `sort_values()` is used to sort Series or DataFrame values in ascending or descending order.

## Syntax
```python
s.sort_values()
df.sort_values(by='column_name')

````python
import numpy as np
import pandas as pd
````

````python
df = pd.read_csv('matches.csv')
df.head()
````

````python
df['team1'].value_counts() + df['team2'].value_counts()
````

````python
(df['team1'].value_counts() + df['team2'].value_counts()).sort_values(ascending=False)
````

````python
(df['team1'].value_counts() + df['team2'].value_counts()).sort_values(ascending=False).head()
````

````python
df.sort_values('city', ascending=False)
````

````python
df
````

### iplace original mai change karta hai

````python
df.sort_values('city', ascending=False, inplace=True)
df
````

### multiple value sort

````python
df.sort_values(['city', 'date'])
````

````python
df.sort_values(['city', 'date'], ascending=[True, False])
````
