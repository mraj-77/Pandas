# fillna()

- `fillna()` is used to fill missing values (NaN) in a Series or DataFrame.
- Mainly used in Data Cleaning and ML Preprocessing.

## Syntax
```python
df.fillna(value)

# Fill all NaN with 0
df.fillna(0)

# Fill with a specific value
df['Age'].fillna(25)

# Fill with mean
df['Age'].fillna(df['Age'].mean())

# Fill with median
df['Age'].fillna(df['Age'].median())

# Fill with mode
df['City'].fillna(df['City'].mode()[0])

# Forward Fill (Previous value)
df.ffill()

# Backward Fill (Next value)
df.bfill()

# Fill specific columns
df.fillna({'Age': 25, 'Salary': 0})

# Modify original DataFrame
df.fillna(0, inplace=True)

Important Points
- fillna() handles missing values.
- Mean → Best for relatively symmetric numerical data.
- Median → Useful when data contains outliers.
- Mode → Useful for categorical data.
- Forward fill → Useful for time-series data.
- Backward fill → Uses next available value.
- Avoid filling missing values before train-test split when learning statistics from data; calculate imputation values using training data only.
Related Methods
df.isnull()          # Check missing valuesdf.isnull().sum()    # Count missing valuesdf.notnull()         # Check non-missing valuesdf.dropna()          # Remove missing valuesdf.fillna(0)         # Fill missing values


Real-Life Use Cases
1. Fill missing student marks with median.
2. Fill missing customer age with median.
3. Fill missing city with mode.
4. Handle missing sales records.
5. Handle missing time-series data.
6. Prepare clean data for Machine Learning.


Difference: dropna() vs fillna()
dropna()	fillna()
Removes missing values	Fills missing values
Can remove rows/columns	Replaces NaN with values
Data may be lost	Data is retained


**Interview Tip:**
- `dropna()` → Remove missing values.
- `fillna()` → Replace missing values.
- In ML, don't blindly drop rows; check how much data will be lost
