# Introduction to Pandas

## What is Pandas?
* **Pandas** Python ki ek open-source library hai jo data manipulation, data cleaning, aur data analysis ke liye use hoti hai (isaka naam "Panel Data" se aaya hai).
* Yeh NumPy ke upar bani hai aur tabular data (rows aur columns) ko handle karna bohot aasan banati hai.
* **Real-life Example:** Jaise Excel spreadsheet mein bade data ko filter, sort, aur analyze karna, bilkul wahi kaam Python mein Pandas aur bhi fast tareeke se karta hai.

---

# Pandas Main Data-Structures

## 1. DataFrames
* Yeh ek **2-Dimensional (2D)** tabular data structure hota hai jisme **rows aur columns** dono hote hain (bilkul Excel sheet ya SQL table ki tarah)[cite: 7].
* Har column mein alag-alag data types (int, float, string) ho sakte hain.
* **Real-life Example:** Jaise ek poore school ka attendance ya marks register jisme Name, Roll No, aur Marks ke alag-alag columns aur rows hoti hain.

## 2. Series
* Yeh ek **1-Dimensional (1D)** labeled array hota hai jo kisi bhi data type (integers, strings, floats) ko store kar sakta hai[cite: 7].
* DataFrame ka har ek single column actually ek **Series** hi hota hai.
* **Real-life Example:** Jaise grocery shopping ki ek single list ya class ke sirf Maths ke marks ka ek akela column.

---

## Important Points ⭐
* **Pandas** = Tabular data analysis aur manipulation ke liye best Python library.
* **DataFrame** = 2D table (Rows + Columns)[cite: 7].
* **Series** = 1D single column with index labels[cite: 7].
