# `isin()` & IPL Death Overs Analysis
* **Task:** Find the most destructive batsman in death overs (overs 16–20) in IPL history.
* **Formula:** $\text{Strike Rate} = \left(\frac{\text{Number of Runs}}{\text{Number of Balls}}\right) \times 100$
* **Condition:** Minimum 200 balls faced in death overs.

```python
import numpy as np
import pandas as pd
```

```python
ipl = pd.read_csv('deliveries.csv')
```

```python
# 1. Filter death overs (overs > 15, i.e., 16th to 20th over)
death_over = ipl[ipl['over'] > 15]
```

```python
# 2. Count total balls faced by each batsman in death overs
all_batsman = death_over.groupby('batsman')['batsman_runs'].size()
all_batsman
```

```python
# 3. Filter batsmen who have faced at least 200 balls in death overs
x = all_batsman >= 200
all_batsman[x]
```

```python
# Extract the list of qualifying batsmen names
ball_200 = all_batsman[x].index.tolist()
len(ball_200)
```

# Calculate Total Runs, Balls Played & Strike Rate of the 43 Players

```python
# 4. Filter dataset for only those 43 players using isin()
mask = ipl['batsman'].isin(ball_200)
final = ipl[mask]
final
```

```python
# 5. Calculate total runs scored by each qualifying batsman
runs = final.groupby('batsman')['batsman_runs'].sum()
```

```python
# 6. Calculate total balls played by each qualifying batsman
balls = final.groupby('batsman')['batsman_runs'].size()
```

```python
# 7. Compute Strike Rate and sort in descending order
sr = (runs / balls) * 100
final_sr = sr.sort_values(ascending=False)
final_sr.sort_values(ascending=False)
```
