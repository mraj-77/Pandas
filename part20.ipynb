# dropna()

- `dropna()` is used to remove missing values (NaN) from a Series or DataFrame.
- Mainly used in Data Cleaning and EDA.

## Syntax
```python
df.dropna()

# Remove rows containing any NaN
df.dropna()

# Remove columns containing NaN
df.dropna(axis=1)

# Remove rows only when all values are NaN
df.dropna(how='all')

# Remove rows if any value is NaN
df.dropna(how='any')

# Check specific columns
df.dropna(subset=['Age'])

# Check multiple columns
df.dropna(subset=['Age', 'Salary'])

# Keep rows having at least 3 non-NaN values
df.dropna(thresh=3)

# Modify original DataFrame
df.dropna(inplace=True)

Real-Life Use Cases
1. Remove rows with missing customer records.
2. Remove columns having excessive missing values.
3. Clean datasets before EDA.
4. Remove incomplete records.
5. Prepare datasets for Machine Learning.

Important Points
- Default axis=0 (rows).
- Default how='any'.
- how='any' → Drop row if any value is NaN.
- how='all' → Drop row if all values are NaN.
- thresh specifies minimum non-null values required.
- Original DataFrame remains unchanged by default.
- inplace=True modifies original DataFrame.

Important Points
- axis=0 → Rows.
- axis=1 → Columns.
- Default axis is 0.
- Original DataFrame remains unchanged by default.
- inplace=True modifies the original DataFrame.
- drop() removes data using index labels, not row positions.

Difference
Method	                Use
drop()	                Remove rows or columns
dropna()	            Remove missing values
drop_duplicates()	    Remove duplicate rows

# Drop single row
df.drop(0)

# Drop multiple rows
df.drop([0, 1, 2])

# Drop single column
df.drop('Age', axis=1)

# Drop multiple columns
df.drop(['Age', 'Salary'], axis=1)

# Using index parameter
df.drop(index=[0, 1])

# Using columns parameter
df.drop(columns=['Age', 'Salary'])

# Modify original DataFrame
df.drop('Age', axis=1, inplace=True)

# Ignore errors if label doesn't exist
df.drop('Age', axis=1, errors='ignore')


Real-Life Use Cases
1. Remove unnecessary columns.
2. Remove unwanted rows.
3. Remove duplicate or irrelevant features after analysis.
4. Clean datasets before Machine Learning.
