# set_index() and reset_index()

- `set_index()` is used to make an existing column the DataFrame index.
- `reset_index()` is used to reset the index back to the default integer index.

## 1. set_index()

### Syntax
```python
df.set_index('column_name')

Important Points
- Default index starts from 0.
- set_index() does not modify original DataFrame by default.
- reset_index() does not modify original DataFrame by default.
- drop=True removes the old index.
- Both methods are useful in Data Cleaning and EDA.
Real-Life Use Cases
1. Set Student_ID as index.
2. Set Date as index in time-series data.
3. Reset index after groupby().
4. Reset index after filtering data.
5. Prepare clean data for Machine Learning.

**Interview Tip:**
```python
df.groupby('City')['Sales'].sum().reset_index()

reset_index() converts the grouped index back into a normal column.
