# virat kohli ne kis team ke khilaf jyada run banaya 10 year mai

```python
vk = ipl[ipl['batsman'] == 'V Kohli']
```

```python
vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head()
```

```python
def run_score(player_name):
    vk = ipl[ipl['batsman'] == player_name]
    return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head().index[0]
```

```python
run_score('V Kohli')
```
