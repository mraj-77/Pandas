# Fetching Rows and Columns in Pandas

## 1. Fetching Columns
* DataFrame mein se kisi specific column ko access karne ke liye column ka naam square brackets `[]` mein diya jata hai.
* **Syntax:** `df['column_name']` (ya dot notation `df.column_name`)
* **Real-life Example:** Jaise ek student database mein se sirf 'Name' ya 'Marks' wale column ko alag nikal lena.

## 2. Fetching Rows using `.loc[]` (Label-based Indexing)
* Rows ko unke **labels ya row index names** ke base par access karne ke liye `.loc[]` ka use hota hai.
* **Syntax:** `df.loc[row_label]` ya `df.loc[start:end, 'column_name']`
* **Real-life Example:** Jaise register mein roll number 'R101' wale student ka poora record nikalna.

## 3. Fetching Rows using `.iloc[]` (Integer position-based Indexing)
* Rows aur columns ko unke **numerical index number** (0-based position) ke base par access karne ke liye `.iloc[]` ka use hota hai.
* **Syntax:** `df.iloc[row_index, column_index]` (jaise `df.iloc[0, 1]` matlab 0th row aur 1st column ki value).
* **Real-life Example:** Jaise line mein khade hue pehle se paanchve number ke student ka data nikalna.

---

## Important Points ⭐
* **Column Access** = `df['col_name']` ka use karke single ya multiple columns nikalna.
* **`.loc[]`** = Label ya name-based row/column selection ke liye.
* **`.iloc[]`** = Integer position (0, 1, 2...) based row/column selection ke liye.
